# Time series foundations
A forecast is defined by *when* it is made and *how far ahead* it predicts. Random shuffling destroys that contract, so all features and splits must respect time.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error

rng=np.random.default_rng(42)
idx=pd.date_range('2022-01-01',periods=900,freq='D')
t=np.arange(len(idx))
y=100+.04*t+12*np.sin(2*np.pi*t/7)+5*np.sin(2*np.pi*t/365.25)+rng.normal(0,4,len(t))
s=pd.Series(y,index=idx,name='demand')
df=s.to_frame()
df['lag_1']=df.demand.shift(1); df['lag_7']=df.demand.shift(7)
df['rolling_7']=df.demand.shift(1).rolling(7).mean()
df.tail()
```

```python
test=df.iloc[-120:].dropna()
mae_lag1=mean_absolute_error(test.demand,test.lag_1)
mae_lag7=mean_absolute_error(test.demand,test.lag_7)
pd.Series({'naive_lag1_MAE':mae_lag1,'seasonal_lag7_MAE':mae_lag7}).round(3)
```

```python
s.iloc[-180:].plot(figsize=(10,4),title='Synthetic daily demand: trend + weekly/annual seasonality'); plt.ylabel('demand'); plt.show()
```

## Baseline first
A sophisticated forecaster that cannot beat a sensible seasonal-naive baseline is not useful. The baseline also reveals the amount of predictable structure that comes from persistence and seasonality alone.

# Feature importance without overclaiming
Impurity importance is convenient but biased toward features with many possible split points. Permutation importance asks a more operational question: how much does validation performance deteriorate if a feature is shuffled?

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance

data=load_breast_cancer(as_frame=True)
Xtr,Xte,ytr,yte=train_test_split(data.data,data.target,test_size=.3,stratify=data.target,random_state=42)
m=RandomForestClassifier(n_estimators=300,min_samples_leaf=2,n_jobs=-1,random_state=42).fit(Xtr,ytr)
imp=pd.Series(m.feature_importances_,index=Xtr.columns,name='impurity')
perm=permutation_importance(m,Xte,yte,n_repeats=10,random_state=42,n_jobs=-1)
out=pd.concat([imp,pd.Series(perm.importances_mean,index=Xtr.columns,name='permutation')],axis=1)
out.sort_values('permutation',ascending=False).head(10).round(4)
```

## Caution
Importance is model- and dataset-dependent. Correlated features can share or mask importance. Neither method establishes causality. I treat importance as a debugging and prioritization signal, not a scientific effect estimate.

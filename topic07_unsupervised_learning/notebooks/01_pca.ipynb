# PCA — compression before decoration
PCA finds orthogonal directions of maximum variance. Because variance depends on units, I scale first unless the original units are intentionally comparable.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X,y=load_wine(return_X_y=True,as_frame=True)
Xs=StandardScaler().fit_transform(X)
pca=PCA().fit(Xs)
cum=np.cumsum(pca.explained_variance_ratio_)
pd.DataFrame({'component':range(1,len(cum)+1),'explained':pca.explained_variance_ratio_,'cumulative':cum}).head(10).round(4)
```

```python
k=int(np.argmax(cum>=.9)+1)
Z=PCA(n_components=k).fit_transform(Xs)
print('components for >=90% variance:',k,'reduced shape:',Z.shape)
```

## Caution
High variance is not synonymous with predictive usefulness. PCA is unsupervised: it does not know the target and can preserve nuisance variance while discarding low-variance predictive signal.

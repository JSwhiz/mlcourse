# Clustering without inventing meaning
Clusters are algorithmic partitions, not discovered natural species. I compare candidate k values using silhouette score and inspect stability before assigning business meaning.

```python
import pandas as pd
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X,_=make_blobs(n_samples=1000,centers=4,cluster_std=[1.0,1.4,.8,1.2],random_state=42)
X=StandardScaler().fit_transform(X)
rows=[]
for k in range(2,8):
    labels=KMeans(n_clusters=k,n_init=20,random_state=42).fit_predict(X)
    rows.append({'k':k,'silhouette':silhouette_score(X,labels)})
pd.DataFrame(rows).round(4)
```

```python
best=max(rows,key=lambda r:r['silhouette'])['k']
km=KMeans(n_clusters=best,n_init=20,random_state=42).fit(X)
pd.DataFrame(km.cluster_centers_,columns=['x1_scaled','x2_scaled']).round(3)
```

## Interpretation
Silhouette measures geometric separation under the chosen distance; it does not validate whether segments are actionable, stable over time or causally different. Those are separate questions.

# Boosting vs Random Forest
The comparison is about ensemble mechanisms, not brand names: random forest averages decorrelated trees in parallel; boosting adds trees sequentially to correct current errors.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier

X,y=load_breast_cancer(return_X_y=True)
cv=StratifiedKFold(5,shuffle=True,random_state=42)
models={
 'random_forest':RandomForestClassifier(n_estimators=300,min_samples_leaf=2,n_jobs=-1,random_state=42),
 'hist_gradient_boosting':HistGradientBoostingClassifier(max_iter=200,learning_rate=.06,max_leaf_nodes=15,l2_regularization=1,random_state=42),
}
rows=[]
for name,m in models.items():
    s=cross_validate(m,X,y,cv=cv,scoring={'auc':'roc_auc','bal_acc':'balanced_accuracy'})
    rows.append({'model':name,'ROC_AUC':s['test_auc'].mean(),'balanced_accuracy':s['test_bal_acc'].mean(),'fit_seconds':s['fit_time'].mean()})
pd.DataFrame(rows).round(4)
```

## Interpretation
One dataset cannot establish a universal winner. I choose using validation quality, latency, memory, calibration/interpretability requirements and operational constraints. Boosting often extracts subtle structure but is more sequential and tuning-sensitive.

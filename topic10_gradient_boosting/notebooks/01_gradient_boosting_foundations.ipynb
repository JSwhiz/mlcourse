# Gradient boosting foundations
Unlike bagging, boosting is sequential: each new weak learner is added to reduce the remaining error. Learning rate controls how aggressively each stage changes the ensemble.

```python
import pandas as pd
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error

X,y=make_regression(n_samples=1800,n_features=20,n_informative=10,noise=30,random_state=42)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=42)
rows=[]
for lr in [.02,.05,.1,.2]:
    m=GradientBoostingRegressor(n_estimators=300,learning_rate=lr,max_depth=2,random_state=42).fit(Xtr,ytr)
    rows.append({'learning_rate':lr,'train_RMSE':mean_squared_error(ytr,m.predict(Xtr))**.5,'test_RMSE':mean_squared_error(yte,m.predict(Xte))**.5})
pd.DataFrame(rows).round(3)
```

```python
m=GradientBoostingRegressor(n_estimators=250,learning_rate=.05,max_depth=2,random_state=42).fit(Xtr,ytr)
curve=[mean_squared_error(yte,p)**.5 for p in m.staged_predict(Xte)]
best_step=int(pd.Series(curve).idxmin()+1)
print('best validation stage:',best_step,'RMSE:',round(min(curve),3))
```

## Takeaway
More estimators are not automatically better. Shrinkage and tree depth determine the pace and shape of corrections; staged validation/early stopping protects against continuing after generalization stops improving.

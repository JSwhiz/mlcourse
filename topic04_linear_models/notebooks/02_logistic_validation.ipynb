# Logistic regression and validation
The model is linear in log-odds, not in predicted probability. Scaling and regularization belong inside the validation pipeline.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score, learning_curve

X,y=load_breast_cancer(return_X_y=True)
cv=StratifiedKFold(5,shuffle=True,random_state=42)
rows=[]
for C in [.01,.1,1,10,100]:
    m=make_pipeline(StandardScaler(),LogisticRegression(C=C,max_iter=5000))
    s=cross_val_score(m,X,y,cv=cv,scoring='roc_auc')
    rows.append({'C':C,'ROC_AUC_mean':s.mean(),'std':s.std()})
pd.DataFrame(rows).round(4)
```

```python
model=make_pipeline(StandardScaler(),LogisticRegression(C=1,max_iter=5000))
sizes,train,val=learning_curve(model,X,y,cv=cv,scoring='roc_auc',train_sizes=[.2,.4,.6,.8,1.0])
pd.DataFrame({'train_size':sizes,'train_auc':train.mean(1),'cv_auc':val.mean(1)}).round(4)
```

## Takeaway
A learning curve distinguishes ‘need a more flexible model’ from ‘need more data’. Regularization controls complexity; the test set should not be used to pick C.

# Regression and regularization
I want to see when coefficient shrinkage improves generalization, especially with correlated/noisy features.

```python
import pandas as pd
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score

X, y = make_regression(n_samples=700, n_features=30, n_informative=8, noise=28, effective_rank=12, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.25, random_state=42)
models = {
 'OLS': LinearRegression(),
 'Ridge': make_pipeline(StandardScaler(), Ridge(alpha=10)),
 'Lasso': make_pipeline(StandardScaler(), Lasso(alpha=1, max_iter=10000)),
}
rows=[]
for name,m in models.items():
    m.fit(X_train,y_train); p=m.predict(X_test)
    rows.append({'model':name,'RMSE':mean_squared_error(y_test,p)**.5,'R2':r2_score(y_test,p)})
pd.DataFrame(rows).round(3)
```

## Interpretation
Regularization is not a magic score booster. It encodes a preference for smaller coefficients and trades some bias for lower variance. Its strength must be selected by validation.
